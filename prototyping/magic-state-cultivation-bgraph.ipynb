```python
blockgraphs = {}
```

```python
blockgraph = BlockGraph.from_bgraph(f"""BLOCKGRAPH 0.1.0;

METADATA: attr_name; value;
source; manual;
circuit_name; cultivation [horizontal];

CUBES: index;x;y;z;kind;label;
0;0;0;2;ZXX;;
1;0;0;3;ZXZ;;
2;0;0;4;ZXX;;
3;0;1;0;ZXt;;
4;0;1;3;ZXX;;
5;0;2;0;ZXt;;
6;0;2;3;ZXX;;

PIPES: src;tgt;kind;
0;1;ZXO;
1;2;ZXO;
1;4;ZOX;
3;4;ZXO;
4;6;ZOX;
5;6;ZXO;
""", graph_name=f"Magic state cultivation")

blockgraphs['horizontal'] = blockgraph
```

```python
blockgraph = BlockGraph.from_bgraph(f"""BLOCKGRAPH 0.1.0;

METADATA: attr_name; value;
source; manual;
circuit_name; cultivation [vertical];

CUBES: index;x;y;z;kind;label;
0;0;0;2;XZX;;
1;0;0;3;XZZ;;
2;0;0;4;XZX;;
3;1;0;0;XZt;;
4;1;0;3;XZX;;
5;2;0;0;XZt;;
6;2;0;3;XZX;;

PIPES: src;tgt;kind;
0;1;XZO;
1;2;XZO;
1;4;OZX;
3;4;XZO;
4;6;OZX;
5;6;XZO;
""", graph_name=f"Magic state cultivation")

blockgraphs['vertical'] = blockgraph
```

```python
for label, blockgraph in blockgraphs.items():
    compiled = compile_block_graph(blockgraph, observables=None)
    circuit = compiled.generate_stim_circuit(k=3, manhattan_radius=0)
    display(Markdown(f"[Open in Crumble ({label}:X)]({compiled.generate_crumble_url(k=3, manhattan_radius=0, add_polygons=True)})"))
```
