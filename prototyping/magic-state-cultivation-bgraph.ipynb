```python
from tqec import Basis
from tqec.computation.block_graph import BlockGraph
from tqec.computation.cube import ZXCube
```

```python
basis = Basis.X
blockgraphs = {}
```

```python
blockgraph = BlockGraph.from_bgraph(f"""BLOCKGRAPH 0.1.0;

METADATA: attr_name; value;
source; manual;
circuit_name; cultivation [horizontal];

CUBES: index;x;y;z;kind;label;
0;0;0;2;PORT;DATA_I;
1;0;0;3;ZXZ;;
2;0;0;4;PORT;DATA_O;
3;0;1;0;ZXt;;
4;0;1;3;ZXX;;

PIPES: src;tgt;kind;
0;1;ZXO;
1;2;ZXO;
1;4;ZOX;
3;4;ZXO;
""", graph_name=f"Magic state cultivation")

blockgraph.fill_ports({
    "DATA_I": ZXCube.from_str(f"ZX{basis}"), "DATA_O": ZXCube.from_str(f"ZX{basis}")
})

blockgraphs['horizontal'] = blockgraph
```

```python
blockgraph.view_as_html()
```

```python
blockgraph = BlockGraph.from_bgraph(f"""BLOCKGRAPH 0.1.0;

METADATA: attr_name; value;
source; manual;
circuit_name; cultivation [vertical];

CUBES: index;x;y;z;kind;label;
0;0;0;2;PORT;DATA_I;
1;0;0;3;XZZ;;
2;0;0;4;PORT;DATA_O;
3;1;0;0;XZt;;
4;1;0;3;XZX;;

PIPES: src;tgt;kind;
0;1;XZO;
1;2;XZO;
1;4;OZX;
3;4;XZO;
""", graph_name=f"Magic state cultivation")

blockgraph.fill_ports({
    "DATA_I": ZXCube.from_str(f"XZ{basis}"), "DATA_O": ZXCube.from_str(f"XZ{basis}")
})

blockgraphs['vertical'] = blockgraph
```
