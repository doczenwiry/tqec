```python
from tqec import Basis
from tqec.computation.block_graph import BlockGraph
from tqec.computation.cube import ZXCube
from tqec.utils.position import Position3D
```

```python
blockgraphs = {}

for basis in [ Basis.X, Basis.Z ]:
    blockgraph = BlockGraph(f"Magic state cultivation [basis={basis}]")

    # Data qubit
    d0, d1, d2, d3, d4 = map(lambda z: Position3D(0,0,z), range(0,5))
    blockgraph.add_cube(d0, "PORT", label="DATA_I")
    blockgraph.add_cube(d1, "XZZ")
    blockgraph.add_cube(d2, "XZZ")
    blockgraph.add_cube(d3, "XZZ")
    blockgraph.add_cube(d4, "PORT", label="DATA_O")
    blockgraph.add_pipe(d0, d1)
    blockgraph.add_pipe(d1, d2)
    blockgraph.add_pipe(d2, d3)
    blockgraph.add_pipe(d3, d4)

    # Cultivation qubit
    s0, s1, s2 = map(lambda z: Position3D(1,0,z), range(2,5))
    blockgraph.add_cube(s0, "XZt")
    blockgraph.add_cube(s1, f"XZX")
    blockgraph.add_pipe(s0, s1, "XZO")

    blockgraph.add_pipe(d3, s1, "OZX")

    blockgraph.fill_ports({ 'DATA_I': ZXCube.from_str(f"XZ{basis}"), 'DATA_O': ZXCube.from_str(f"XZ{basis}") })

    blockgraphs[basis] = blockgraph
```
