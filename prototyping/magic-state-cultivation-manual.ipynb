```python
from IPython.core.display import Markdown
from IPython.display import display

from tqec import compile_block_graph
from tqec.computation.block_graph import BlockGraph
from tqec.utils.position import Position3D
```

```python
blockgraphs = {}
```

```python
blockgraph = BlockGraph(f"Magic state cultivation")

# Data qubit
d0, d1, d2, d3, d4 = map(lambda z: Position3D(0,0,z), range(0,5))
blockgraph.add_cube(d0, "XZX")
blockgraph.add_cube(d1, "XZZ")
blockgraph.add_cube(d2, "XZZ")
blockgraph.add_cube(d3, "XZZ")
blockgraph.add_cube(d4, "XZX")
blockgraph.add_pipe(d0, d1)
blockgraph.add_pipe(d1, d2)
blockgraph.add_pipe(d2, d3)
blockgraph.add_pipe(d3, d4)

# Cultivation qubit #1
s0, s1 = map(lambda z: Position3D(1,0,z+2), range(2))
blockgraph.add_cube(s0, "XZt")
blockgraph.add_cube(s1, f"XZX")
blockgraph.add_pipe(s0, s1, "XZO")

# Cultivation qubit #2
s2, s3 = map(lambda z: Position3D(2,0,z+2), range(2))
blockgraph.add_cube(s2, "XZt")
blockgraph.add_cube(s3, f"XZX")
blockgraph.add_pipe(s2, s3, "XZO")

blockgraph.add_pipe(d3, s1, "OZX")
blockgraph.add_pipe(s1, s3, "OZX")

blockgraphs['horizontal'] = blockgraph
```

```python
blockgraph.view_as_html()
```

```python
blockgraph = BlockGraph(f"Magic state cultivation")

# Data qubit
d0, d1, d2, d3, d4 = map(lambda z: Position3D(0,0,z), range(0,5))
blockgraph.add_cube(d0, "ZXX")
blockgraph.add_cube(d1, "ZXZ")
blockgraph.add_cube(d2, "ZXZ")
blockgraph.add_cube(d3, "ZXZ")
blockgraph.add_cube(d4, "ZXX")
blockgraph.add_pipe(d0, d1)
blockgraph.add_pipe(d1, d2)
blockgraph.add_pipe(d2, d3)
blockgraph.add_pipe(d3, d4)

# Cultivation qubit #1
s0, s1 = map(lambda z: Position3D(0,1,z+2), range(2))
blockgraph.add_cube(s0, "ZXt")
blockgraph.add_cube(s1, f"ZXX")
blockgraph.add_pipe(s0, s1, "ZXO")

# Cultivation qubit #2
s2, s3 = map(lambda z: Position3D(0,2,z+2), range(2))
blockgraph.add_cube(s2, "ZXt")
blockgraph.add_cube(s3, f"ZXX")
blockgraph.add_pipe(s2, s3, "ZXO")

blockgraph.add_pipe(d3, s1, "ZOX")
blockgraph.add_pipe(s1, s3, "ZOX")

blockgraphs['vertical'] = blockgraph
```

```python
blockgraph.view_as_html()
```

```python
for label, blockgraph in blockgraphs.items():
    compiled = compile_block_graph(blockgraph, observables=None)
    circuit = compiled.generate_stim_circuit(k=3, manhattan_radius=0)
    display(Markdown(f"[Open in Crumble ({label}:X)]({compiled.generate_crumble_url(k=3, manhattan_radius=0, add_polygons=True)})"))
```
